# 03 — Data Quality

## Objective

Test the dataset without assuming that unusual values are errors.

## Quality dimensions

- Completeness — اكتمال البيانات — Complétude
- Uniqueness — التفرد — Unicité
- Referential Integrity — سلامة العلاقات المرجعية — Intégrité référentielle
- Validity — الصلاحية — Validité
- Range Validation — التحقق من النطاق — Validation des plages
- Accepted Values — القيم المقبولة — Valeurs autorisées
- Temporal Consistency — الاتساق الزمني — Cohérence temporelle
- Volume Checks — فحوصات حجم البيانات — Contrôles de volume
- Outlier Detection — اكتشاف القيم الشاذة — Détection des valeurs aberrantes

## Analyst question

Should a NULL, repeated ID or unusual value be deleted?

**Answer:** Not automatically.

First determine why it exists, whether it is allowed by the official definition, and whether it affects a specific metric or the dataset globally.

Every quality issue receives a documented decision.

## Quality result interpretation

For every failed or unusual check, record:

1. issue;
2. evidence;
3. likely explanation;
4. impact;
5. action;
6. whether the original data remains untouched.
