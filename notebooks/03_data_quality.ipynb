# 03 — Data Quality## ObjectiveTest the dataset without assuming that unusual values are errors.## Quality dimensions- Completeness — Complétude- Uniqueness — Unicité- Referential Integrity — Intégrité référentielle- Validity — Validité- Range Validation — Validation des plages- Accepted Values — Valeurs autorisées- Temporal Consistency — Cohérence temporelle- Volume Checks — Contrôles de volume- Outlier Detection — Détection des valeurs aberrantes

## Analyst question

Should a NULL, repeated ID or unusual value be deleted?

**Answer:** Not automatically.

First determine why it exists, whether it is allowed by the official definition, and whether it affects a specific metric or the dataset globally.

Every quality issue receives a documented decision.

## Quality result interpretation

For every failed or unusual check, record:

1. issue;
2. evidence;
3. likely explanation;
4. impact;
5. action;
6. whether the original data remains untouched.
