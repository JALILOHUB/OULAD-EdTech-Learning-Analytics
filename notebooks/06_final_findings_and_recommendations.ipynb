## Recommendation rule

Every recommendation must trace back to evidence.

A recommendation should answer:

- What was observed?
- Why does it matter to the client?
- What action could be considered?
- What evidence supports it?
- What uncertainty remains?

## Final quality gate

Before publishing:

- [ ] All important claims have evidence.
- [ ] SQL results are reproducible.
- [ ] Data-quality decisions are documented.
- [ ] Unexpected results are not hidden.
- [ ] Limitations are explicit.
- [ ] Recommendations are evidence-based.
- [ ] No causal claim is made from association alone.
