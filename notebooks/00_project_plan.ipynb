### Why this architecture?

SQL is the backbone because the dataset is relational and the project must demonstrate professional SQL analytics.

Python is used where it is stronger: ingestion, profiling and advanced statistical analysis.

Power BI is used for interactive communication after the analytical layer exists.

## Initial business problem> How can an online learning platform identify engagement patterns, academic risk, and factors associated with student success?The analysis must remain careful about causality. Use **associated with** for descriptive relationships unless the evidence and research design support causal inference.
