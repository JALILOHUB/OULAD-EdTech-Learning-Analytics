## Project architecture

Official Data Source  
↓  
Python Ingestion  
↓  
PostgreSQL RAW Layer  
↓  
Data Quality  
↓  
PostgreSQL STAGING Layer  
↓  
SQL Analytical Layer / Data Mart  
↓  
Power BI  
↓  
Python Advanced Analytics when justified

### Why this architecture?

SQL is the backbone because the dataset is relational and the project must demonstrate professional SQL analytics.

Python is used where it is stronger: ingestion, profiling and advanced statistical analysis.

Power BI is used for interactive communication after the analytical layer exists.

## Initial business problem

> How can an online learning platform identify engagement patterns, academic risk, and factors associated with student success?

The analysis must remain careful about causality.

We use **associated with — مرتبط بـ — associé à** unless the evidence supports causal inference.
