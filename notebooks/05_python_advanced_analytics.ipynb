# 05 — Python Advanced Analytics

## Objective

Use Python only where it adds analytical value beyond efficient SQL.

Potential work:
- exploratory data analysis;
- statistical analysis;
- distribution analysis;
- outlier analysis;
- segmentation;
- feature engineering;
- machine learning if a justified business question remains.

## Analyst question

Why are we using Python here instead of SQL?

**Answer must be written before the code.**

Examples of valid reasons:
- statistical test;
- complex numerical transformation;
- visualization not better handled in Power BI;
- model training;
- specialized analytical library.

## Result interpretation

Explain the output in simple professional English.

Separate:
- descriptive finding;
- association;
- predictive performance;
- limitation.

Never convert a correlation or prediction into a causal claim.
