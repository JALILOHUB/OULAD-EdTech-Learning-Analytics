# 00 Environment Test

## Objective

Verify that the Python environment can connect to the local PostgreSQL database.

## Expected result

Python should connect successfully to the `oulad_analytics` database and return the current database name and PostgreSQL version.

```python
import os
import psycopg
from dotenv import load_dotenv

# 1. Load environment variables from the .env file
load_dotenv()

# 2. Retrieve database credentials securely
db_host = os.getenv("DB_HOST", "localhost")
db_port = os.getenv("DB_PORT", "5432")
db_name = os.getenv("DB_NAME", "oulad_analytics")
db_user = os.getenv("DB_USER", "postgres")
db_password = os.getenv("DB_PASSWORD")

# 3. Security Check: Ensure password is set and not the default placeholder
if not db_password or db_password == "CHANGE_ME":
    raise ValueError(
        "SECURITY ALERT: DB_PASSWORD is not set or is still 'CHANGE_ME'. "
        "Please copy .env.example to .env and set your actual password."
    )

# 4. Attempt secure connection
try:
    connection = psycopg.connect(
        host=db_host,
        port=db_port,
        dbname=db_name,
        user=db_user,
        password=db_password
    )
    
    cursor = connection.cursor()
    
    cursor.execute("SELECT current_database();")
    database_name = cursor.fetchone()[0]
    
    cursor.execute("SELECT version();")
    postgres_version = cursor.fetchone()[0]
    
    print("✅ Database:", database_name)
    print("✅ PostgreSQL:", postgres_version)
    print("✅ Status: Secure connection successful via environment variables.")
    
    cursor.close()
    connection.close()
    
except Exception as e:
    print(f"❌ Connection failed with error: {e}")
```
